```python
import os
```

```python
results = []
for conf in confs:
    print("*"*80)
    print(conf)
    print("*"*80)

    for v in confs[conf]:
        if conf != "prodzanz8" and conf != "wiki":
            f = f'old_mem_curve/mem_curve_eval_{conf}_{v}.csv'
        else:
            f = f'mem_curve_eval_{v}.csv'
        if not os.path.exists(f):
            print(f, "not ready")
            continue
        res = pd.read_csv(f)
        res = res.drop_duplicates(subset=['data_name', 'batch_id', 'data_batch_id'], keep='first')
        res = res[['data_name', 'batch_id', 'data_batch_id', 'accuracy']]
        res = res.set_index(['data_name', 'batch_id', 'data_batch_id']).sort_index()
        res = res.rename(columns={'accuracy': f'{conf}-{v}-accuracy'})
        results.append(res)
```
