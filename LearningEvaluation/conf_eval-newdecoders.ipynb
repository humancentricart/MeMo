```python
import pandas as pd
```

```python
lhs = {4:[4], 5:[4]}


confs = {
    "wiki":[],
}
for l in lhs:
    for h in lhs[l]:
        confs["wiki"].append(f"wiki_20_6_sumshuffnormprod{h}|{l}_new_decoders_V2")
        confs["wiki"].append(f"wiki_20_6_SCC{h}|{l}_new_decoders_V2")
    
```

```python
import os
```

```python
results = []
for conf in confs:
    print("*"*80)
    print(conf)
    print("*"*80)

    for v in confs[conf]:
        f = f'results_mem_curve_eval/mem_curve_eval_{v}.csv'
        if not os.path.exists(f):
            print(f, "not ready")
            continue
        res = pd.read_csv(f)
        
        res = res[['data_name', 'batch_id', 'data_batch_id'] + [c for c in res.columns if '/accuracy' in c]]
        res = res.set_index(['data_name', 'batch_id', 'data_batch_id']).sort_index()
        res = res.drop_duplicates()
        res = res.rename(columns=lambda x: f'{v} {x}')
        results.append(res)
```

```python
pd.set_option('display.max_rows', None)

pd.concat(results, axis=1).sort_index()#.dropna()
```

```python
(pd.concat(results, axis=1).sort_index()*100).round(2).to_csv('conf_eval_decoders.csv')
```
